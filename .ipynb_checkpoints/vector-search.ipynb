```python
from pathlib import Path
from urllib.request import urlopen

import pandas as pd
import torch
from langchain_chroma import Chroma
from langchain_core.documents import Document
from langchain_huggingface import HuggingFaceEmbeddings
```

```python
import os
import torch


thread_count = max(1, os.cpu_count() or 1)
torch.set_num_threads(thread_count)
torch.set_num_interop_threads(max(1, min(thread_count, 8)))

print(f"Threads da CPU: {thread_count}")
```

```python
CSV_URL = "https://aqui-ta-bom-api.onrender.com/v1/api/postagens/csv"
CSV_PATH = Path("postagens.csv")

with urlopen(CSV_URL, timeout=60) as response:
    CSV_PATH.write_bytes(response.read())

postagens = pd.read_csv(CSV_PATH, encoding="utf-8-sig")
print(f"CSV baixado em: {CSV_PATH.resolve()}")
print(f"Postagens carregadas: {len(postagens)}")
```

```python
postagens.head()
```

```python
postagens[["titulo", "descricao", "nomeRestaurante", "nota", "status"]]
```

```python
# O texto usado na busca combina o conteúdo e o contexto de cada postagem.
postagens["texto_busca"] = (
    postagens[["titulo", "descricao", "nomeRestaurante", "status"]]
    .fillna("")
    .astype(str)
    .agg(" | ".join, axis=1)
)

postagens["texto_busca"].to_csv(
    "postagens_texto.txt",
    index=False,
    header=False,
    lineterminator="\n",
)
```

```python
documents = []
metadata_columns = [
    "id", "usuarioId", "titulo", "imagemUrl", "likes", "nota", "status",
    "nomeUsuario", "nomeRestaurante", "dataCriacao", "quantidadeComentarios",
    "nomesUsuariosCurtiram",
]

for _, postagem in postagens.dropna(subset=["texto_busca"]).iterrows():
    metadata = {
        column: str(postagem[column])
        for column in metadata_columns
        if column in postagens.columns and pd.notna(postagem[column])
    }
    documents.append(
        Document(
            page_content=str(postagem["texto_busca"]),
            metadata=metadata,
        )
    )

print(f"Documentos preparados: {len(documents)}")
```

```python
documents[0]
```

```python
embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2",
    model_kwargs={"device": "cpu"},
    encode_kwargs={
        "normalize_embeddings": True,
        "batch_size": 64,
    },
)

# Os embeddings das postagens são calculados localmente.
db_postagens = Chroma.from_documents(
    documents=documents,
    embedding=embeddings,
    collection_name="postagens",
    persist_directory="db_postagens",
    ids=[document.metadata.get("id", str(index)) for index, document in enumerate(documents)],
)

print("Banco vetorial criado em ./db_postagens")
```

```python
# Para reutilizar o banco já criado sem recalcular os embeddings:
db_postagens = Chroma(
    collection_name="postagens",
    persist_directory="db_postagens",
    embedding_function=embeddings,
)
```

```python
print("PyTorch:", torch.__version__)
print("GPU disponível:", torch.cuda.is_available())
print("Dispositivo dos embeddings: CPU")
```

```python
query = "comida boa e barata para almoçar"
results = db_postagens.similarity_search(query, k=5)

for position, document in enumerate(results, start=1):
    metadata = document.metadata
    print(f"\n{position}. {metadata.get('titulo', 'Título desconhecido')}")
    print(f"   Restaurante: {metadata.get('nomeRestaurante', 'Desconhecido')}")
    print(f"   Nota: {metadata.get('nota', 'N/A')}")
    print(f"   Status: {metadata.get('status', 'N/A')}")
    print(f"   Descrição: {metadata.get('descricao', document.page_content)}")
    print(f"   Imagem: {metadata.get('imagemUrl', 'N/A')}")
```
